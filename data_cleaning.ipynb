```python
# Feature Selection
```

```python
## Create columns like month , year , season

df['month']=df['date'].dt.month
df['year']=df['date'].dt.year

df['season']=df['month'].map({
    12:'Winter',1:'Winter',2:'Winter',
    3:'Summer',4:'Summer',5:'Summer',
    6:'Monsoon',7:'Monsoon',8:'Monsoon',9:'Monsoon',
    10:'Post-Monsoon',11:'Post-Monsoon'
})
```

```python
## Seasonal AQI

seasonal_aqi=df.groupby('season')['us_aqi'].mean()

seasonal_aqi
```

```python
## PM2.5 vs US AQI

plt.figure(figsize=(8,6))

plt.scatter(df['pm2_5'],df['us_aqi'],alpha=0.5)

plt.title("PM2.5 vs US AQI")
plt.xlabel("PM2.5")
plt.ylabel("US AQI")

plt.tight_layout()
plt.show()
```

```python
## data repare for model to predict 

features=['pm10','pm2_5','carbon_monoxide','nitrogen_dioxide','sulphur_dioxide','ozone','aerosol_optical_depth','dust','uv_index']
```

```python
## Separate target and features column

X=df[features]
Y=df['us_aqi']
```

```python
X
```

```python
Y
```

```python
# Separate data for training and testing purpose

from sklearn.model_selection import train_test_split

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
# First Model ---> Linear Regression

from sklearn.linear_model import LinearRegression

lr=LinearRegression()
```

```python
lr.fit(X_train,Y_train)

y_predict=lr.predict(X_test)
```

```python
### Model performance

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mae=mean_absolute_error(Y_test,y_predict)  # prediction average how many AQI points are wrong
mse=mean_squared_error(Y_test,y_predict)   
rmse=np.sqrt(mse)
r2=r2_score(Y_test,y_predict)
```

```python
print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("r2_Score: ",r2)
```

```python
## Second model ----> Random Forest Regressor

from sklearn.ensemble import RandomForestRegressor

rf=RandomForestRegressor(n_estimators=200,random_state=42)

rf.fit(X_train,Y_train)
```

```python
y_pred_rf=rf.predict(X_test)
```

```python
# Performance

mae_rf=mean_absolute_error(Y_test,y_pred_rf)
mse_rf=mean_squared_error(Y_test,y_pred_rf)
rmse_rf=np.sqrt(mse_rf)
r2_rf=r2_score(Y_test,y_pred_rf)
```

```python
print("Random Forest Performance")
print("MAE: ",mae_rf)
print("MSE: ",mse_rf)
print("RMSE: ",rmse_rf)
print("R2_score: ",r2_rf)
```

```python
# Third model ---> Gradient Boosting

from sklearn.ensemble import GradientBoostingRegressor

gb=GradientBoostingRegressor(n_estimators=200,learning_rate=0.05,max_depth=3,random_state=42)

gb.fit(X_train,Y_train)
```

```python
y_pred_gb=gb.predict(X_test)
```

```python
# Performance of GB

mae_gb=mean_absolute_error(Y_test,y_pred_gb)
mse_gb=mean_squared_error(Y_test,y_pred_gb)
rmse_gb=np.sqrt(mse_gb)
r2_gb=r2_score(Y_test,y_pred_gb)
```

```python
print("Gradient Boosting Regressor")
print("MAE: ",mae_gb)
print("MSE: ",mse_gb)
print("RMSE: ",rmse_gb)
print("R2_Score: ",r2_gb)
```

```python
## Graphical Representation

plt.figure(figsize=(8,6))
plt.scatter(Y_test,y_pred_gb,alpha=0.6)

plt.xlabel("Actual US AQI")
plt.ylabel("Predicted US AQI")
plt.title("Actual vs Predicted US AQI - Gradient Boosting")

plt.tight_layout()
plt.show()
```
